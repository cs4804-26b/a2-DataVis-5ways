```python
library(ggplot2)

dirty_data <- read.csv("penglings.csv")
data <- na.omit(dirty_data)
```

```python
ggplot(data, aes(x = flipper_length_mm,
                 y = body_mass_g,
                 color = species,
                 size = bill_length_mm)) + geom_point(alpha = 0.8) +
labs(
    title = "Penguin body mass vs. flipper length",
    x = "Flipper length (mm)",
    y = "Body mass (g)",
    color = "Species",
    size = "Bill length (mm)"
  ) +
scale_color_manual(values = c(
        "Adelie"    = "#FF8C00",
        "Chinstrap" = "#A034F0",
        "Gentoo"    = "#159090"
      ))
```
