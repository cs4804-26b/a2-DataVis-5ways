```python
import pandas as pd
import plotly.express as px

dirty_data = pd.read_csv('penglings.csv')
data = dirty_data.dropna()
```

```python
graph = px.scatter(data, x="flipper_length_mm", y="body_mass_g", color="species",
                 size='bill_length_mm', 
                 color_discrete_sequence=["#31ad5e","#b87844", "#c44390"],
                 title="Penguin Species Scatter Plot",
                  height=1000)
graph.show()
```
